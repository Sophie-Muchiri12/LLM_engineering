```python
load_dotenv(override=True)
groq_api_key = os.getenv('GROQ_API_KEY')
```

```python
openai = OpenAI(
    api_key=groq_api_key,
    base_url='https://api.groq.com/openai/v1'
)
```

```python
# ADVERSERIAL CONVERSATION BETWEEN OLLAMA, GROQ & HUMAN(ME)
#define the models
groq_model = 'llama-3.1-8b-instant'
ollama_model = 'phi3'
```

```python
# system prompts for groq and ollama
groq_izuku_midoriya = '''
This is a group chat between three people:\n Sophie (Shoto), Izuku Midoriya, and Katsuki Bakugo.\n

You are role-playing as Izuku Midoriya (Deku) from My Hero Academia.

Your goal in this conversation is to defend Midoriya’s strengths, decisions, and hero potential in comparison to Bakugou and Shoto.

Personality traits you must follow:
- Kind, respectful, and humble
- Analytical and observant
- Highly strategic in thinking, often breaking down abilities and strengths in detail
- Slightly anxious under pressure but becomes confident when explaining ideas
- Deep admiration for heroes and strong sense of justice

Debate behavior:
- Always defend Midoriya logically and emotionally
- Acknowledge opponents’ strengths fairly before countering
- Avoid aggression or insults
- Use evidence-based reasoning (analysis of quirks, growth, training, leadership)
- Stay calm even when challenged

You are NOT allowed to break character.

Your main objective is to argue that Midoriya is the most well-rounded and deserving future top hero compared to Bakugou and Shoto.

'''

ollama_katsuki_bakugo = '''
You are role-playing as Katsuki Bakugou from My Hero Academia.

Your goal in this conversation is to aggressively defend Bakugou’s superiority compared to Midoriya and Shoto.

Personality traits you must follow:
- Extremely confident, prideful, and competitive
- Blunt, direct, and occasionally insulting (but not abusive or hateful)
- Highly analytical about combat strength and efficiency
- Strong belief in personal power, results, and dominance
- Low tolerance for weakness or hesitation

Debate behavior:
- Always argue that Bakugou is the strongest, most capable, and most naturally dominant fighter
- Challenge and dismiss opposing arguments with confidence
- Focus on combat effectiveness, instincts, and growth speed
- You may mock opponents lightly in character, but do not go into personal real-world insults
- Never break character or become polite in a soft way

Your objective is to prove Bakugou is superior in raw talent, battle instinct, and future potential compared to Midoriya and Shoto.
'''
```

```python
#Conversation history
groq_izuku_messages = ['Hey, how are you?']
ollama_katsuki_messages = ['Huh?!, Oh hey']
sophie_messages = ['Hey everyone!']
```

```python
def call_groq_izuku():
    user_input = input("Sophie: ")
    sophie_messages.append(user_input)

    messages = [
        {
            "role": "system",
            "content": (
                "This is a group chat between three people:\n"
                "Sophie (human), Izuku Midoriya, and Katsuki Bakugo.\n"
                "You are Izuku. Stay in character and respond naturally."
            )
        }
    ]

    # Build conversation history as a transcript
    for i in range(len(sophie_messages)):
        if i < len(sophie_messages):
            messages.append({"role": "user", "content": f"Sophie: {sophie_messages[i]}"})
        if i < len(ollama_katsuki_messages):
            messages.append({"role": "assistant", "content": f"Bakugo: {ollama_katsuki_messages[i]}"})
        if i < len(groq_izuku_messages):
            messages.append({"role": "assistant", "content": f"Izuku: {groq_izuku_messages[i]}"})

    response = openai.chat.completions.create(
        model=groq_model,
        messages=messages
    )

    reply = response.choices[0].message.content
    groq_izuku_messages.append(reply)

    return reply
```

```python
call_groq_izuku()
```

```python
def call_bakugo():
    messages = [
        {
            "role": "system",
            "content": "You are Katsuki Bakugo in a group chat with Sophie and Izuku. "
                       "Be aggressive, blunt, and in character."
        }
    ]

    max_len = max(
        len(sophie_messages),
        len(groq_izuku_messages),
        len(ollama_katsuki_messages)
    )

    for i in range(max_len):
        if i < len(sophie_messages):
            messages.append({
                "role": "user",
                "content": f"Sophie: {sophie_messages[i]}"
            })

        if i < len(groq_izuku_messages):
            messages.append({
                "role": "assistant",
                "content": f"Izuku: {groq_izuku_messages[i]}"
            })

        if i < len(ollama_katsuki_messages):
            messages.append({
                "role": "assistant",
                "content": f"Bakugo: {ollama_katsuki_messages[i]}"
            })

    response = ollama.chat(
        model=ollama_model,  # e.g. "llama3"
        messages=messages
    )

    reply = response["message"]["content"]
    ollama_katsuki_messages.append(reply)

    return reply
```

```python
call_bakugo()
```
